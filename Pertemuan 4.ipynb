```python
total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
tugas = int(input("Masukkan nilai Tugas: "))
uts = int(input("Masukkan nilai UTS: "))
uas = int(input("Masukkan nilai UAS: "))
kuis = int(input("Masukkan nilai Kuis: "))
proyek = int(input("Masukkan nilai Proyek: "))

nilai_mahasiswa = (tugas * 0.3) + (uts * 0.3) + (uas * 0.4) + (kuis * 0.2) + (proyek * 0.4)

standard_kelulusan = int(input("Masukkan standar kelulusan: "))

apakah_lulus = nilai_mahasiswa >= standard_kelulusan
perlu_remidial = nilai_mahasiswa < standard_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standard_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remidial? {perlu_remidial}")
```

```python
belanja = int(input("Masukkan total belanja: "))
diskon = (0, 0.35 * belanja)[belanja > 1000000]
total_harga = belanja - diskon
print("Harga setelah diskon =", total_harga)
```

```python
tahun = int(input("Masukkan tahun: "))
kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
print(f"Tahun {tahun} adalah tahun kabisat? {kabisat}")
```

```python
harga_per_kg = int(input("Masukkan harga daging ikan giling per kg: "))
jumlah_kg = int(input("Masukkan jumlah kg yang ingin dibeli: "))
uang_bayar = int(input("Masukkan jumlah uang yang dibayar: "))

total_harga = harga_per_kg * jumlah_kg
kembalian = uang_bayar - total_harga
uang_kurang = uang_bayar < total_harga

print(f"Total harga yang harus dibayar: Rp {total_harga}")
print(f"Kembalian transaksi: Rp {kembalian}")
print(f"Uangnya kurang! {uang_kurang}")
```
