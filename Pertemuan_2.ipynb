```python
# Latihan 1
# Membuat tiga variabel yang sesuai dengan konvensi PEP 8

tegangan = 18                      # int = volt
arus = 6                           # int = ampere
jenis_sel = "Monocrystalline"      # str = teks

# Cetak ketiganya
print(tegangan, arus, jenis_sel)
```

```python
# Latihan 2
# Satu variabel untuk tiap tipe data primitif konteks komponen elektronika

jumlah_resistor = 12              # int   = jumlah komponen (bilangan bulat)
nilai_resistor = 4.7              # float = hambatan dalam kilo-ohm (desimal)
komponen_aktif = True             # bool  = apakah komponen aktif
nama_komponen = "Kapasitor"       # str   = nama komponen (teks)

# Nilai dan tipe tiap variabel
print(jumlah_resistor, type(jumlah_resistor))
print(nilai_resistor, type(nilai_resistor))
print(komponen_aktif, type(komponen_aktif))
print(nama_komponen, type(nama_komponen))
```

```python
# Latihan 3
# Menentukan tipe data yang paling sesuai lalu membuat variabelnya

nama_unit = "PLTS Nunukan"        # str   = teks (nama unit)
jumlah_panel = 240                # int   = jumlah panel (bilangan bulat)
tegangan_keluaran = 48.5          # float = tegangan (desimal)
sedang_beroperasi = True          # bool  = status logika (True/False)

# Buktikan dengan type()
print(nama_unit, "->", type(nama_unit))
print(jumlah_panel, "->", type(jumlah_panel))
print(tegangan_keluaran, "->", type(tegangan_keluaran))
print(sedang_beroperasi, "->", type(sedang_beroperasi))
```

```python
# Latihan 4
# Data:
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir.
print("Pembacaan pertama:", tegangan_harian[0])    # indeks 0 = pertama
print("Pembacaan terakhir:", tegangan_harian[5])   # indeks 5 = terakhir

# (b) Tambah satu nilai baru.
tegangan_harian.append(222.0)

# (c) Cetak jumlah total.
print("Jumlah total data:", len(tegangan_harian))
```

```python
# Latihan 5
# Dictionary spesifikasi motor listrik

motor = {
    "nama": "Motor Induksi 3 Fasa",
    "daya_kW": 37,
    "efisiensi": 92.7,
    "frequensi": 50,
    "tegangan_V": 380,
    "rpm": 2900
    }

# Cetak salah satu nilai lewat key.
print("Daya motor:", motor["daya_kW"], "kW")

# Ubah nilai rpm.
motor["rpm"] = 2945

# Cetak ulang seluruh dictionary.
print(motor)
```

```python
# Latihan 6
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 50, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 150,  "status": "pemeliharaan"},
]

# (a) Cetak nama unit kedua.
print("Nama unit kedua:", pembangkit[1]["nama"])

# (b) Ubah status salah satu unit.
pembangkit[1]["status"] = "pemeliharaan"

# (c) Tambahkan satu unit baru ke list dengan append().
pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 200, "status": "operasi"})

# (d) Cetak seluruh data.
print(pembangkit)
```
