```python
gorengan_rani = 3 * 2500
esteh_rani = 2 * 4000
uangbayar = 20000

total_belanja = gorengan_rani + esteh_rani
kembalian = uangbayar - total_belanja
uang_cukup = uangbayar >= total_belanja

print(f"Apakah uang Rani cukup?: {uang_cukup}")
print(f"Total belanja: {gorengan_rani + esteh_rani}")
print(f"Uang kembalian: {kembalian}")
```

```python
Buku_beli = 7
Harga_buku = 18500
Diskon_total = 10
Uang_bayar = 150000

total_belanja = Buku_beli * Harga_buku
Diskon = total_belanja * (Diskon_total / 100)
Total_bayar = total_belanja - Diskon
patungan = Total_bayar // 4
kurang_patungan = Total_bayar - patungan * 4
uang_cukup = Uang_bayar >= Total_bayar

print(f"Total belanja: {total_belanja}")
print(f"Diskon total: {Diskon}")
print(f"Total bayar: {Total_bayar}")
print(f"Kembalian: {Uang_bayar - Total_bayar}")
print(f"Harga jika patungan: {patungan}")
print(f"Uang Budi cukup: {uang_cukup}")
print(f"Kurang sisa patungan: {kurang_patungan}")
```

```python
jumlah_panel = 12
daya_perpanel = 450
jam_penyinaran = 5
efektivitas = 0.8
kebutuhan = 25
tarif = 1444.70

energi_harian = jumlah_panel * daya_perpanel * jam_penyinaran * efektivitas
energi_harian_kwh = energi_harian / 1000
energi_bulanan = energi_harian_kwh * 30
energihasil_rupiah = energi_bulanan * tarif
kekurangan = kebutuhan - energi_harian_kwh
cukup = energi_harian >= kebutuhan

print(f"Energi harian per kwh: {energi_harian_kwh}")
print(f"Energi bulanan: {energi_bulanan}")
print(f"Energi hasil dalam rupiah: {round(energihasil_rupiah)}")
print(f"Kekurangan energi: {kekurangan:.2f}")
print(f"Apakah cukup?: {cukup}")
```

```python
listrik_generator = 200
jam_operasi = 8
konsumsi_bensin = 0.25
harga_bensin = 13500
kapasitas_generator = 1000
arus_kabel = 20
resistansi = 0.5

energi_perhari_kwh = listrik_generator * jam_operasi
kebutuhan_bensin = energi_perhari_kwh * konsumsi_bensin
harga_bensin_rupiah = harga_bensin * kebutuhan_bensin
lama_tangki_penuh = kapasitas_generator
tangki_penuh_lamahari = 1000 // 400
sisa_bensin = 1000 % 400
rugi_daya = arus_kabel ** 2 * resistansi
rugi_tinggi = rugi_daya >= 150 and arus_kabel >= 20

print(f"Energi harian: {energi_perhari_kwh} kWh")
print(f"Solar per hari: {kebutuhan_bensin} liter")
print(f"Biaya per hari: Rp {harga_bensin_rupiah}")
print(f"Tangki penuh untuk: {tangki_penuh_lamahari} hari")
print(f"Sisa solar: {sisa_bensin} liter")
print(f"Rugi daya kabel: {rugi_daya} W")
print(f"Rugi kabel tergolong tinggi: {rugi_tinggi}")
```
