```python
# Nama   : M. Fattahillah S.
# NIM    : 4232501037
# Tanggal: 29 September 2026
# Tujuan : Menampilkan hasil operasi +, -, *, /, //, dan % dari dua bilangan bulat

a = int(input("Masukkan bilangan bulat pertama: "))
b = int(input("Masukkan bilangan bulat kedua: "))

print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")    # error jika b = 0
print(f"{a} // {b} = {a // b}")   # error jika b = 0
print(f"{a} % {b} = {a % b}")    # error jika b = 0
```

```python
# Nama   : M. Fattahillah S.
# NIM    : 4232501037
# Tanggal: 29 September 2026
# Tujuan : Menentukan apakah seseorang boleh mengemudi sendiri memakai operator logika dan tanda kurung

umur = int(input("Masukkan umur: "))
punya_sim = input("Punya SIM? (True/False): ") == "True"

boleh_mengemudi = umur >= 17 and punya_sim
print(f"Boleh mengemudi sendiri: {boleh_mengemudi}")

izin_khusus = True

tanpa_kurung = umur >= 17 and punya_sim or izin_khusus
dengan_kurung = umur >= 17 and (punya_sim or izin_khusus)

print(f"Tanpa kurung: {tanpa_kurung}")
print(f"Dengan kurung: {dengan_kurung}")

```

```python
# Nama   : M. Fattahillah S.
# NIM    : 4232501037
# Tanggal: 29 September 2026
# Tujuan : Menghitung efisiensi termal unit pembangkit dari energi input dan output (MJ)

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%")
```
